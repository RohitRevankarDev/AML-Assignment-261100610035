# Gaussian Mixture Model (GMM) Clustering

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, adjusted_rand_score

print("Libraries imported successfully.")
```

```python

data = load_iris(as_frame=True)

X = data.data
y = data.target

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Dataset shape:", X.shape)
print("Number of features:", X.shape[1])
print("Features:", list(X.columns))
print("Target classes:", list(data.target_names))
```

```python

gmm = GaussianMixture(
    n_components=3,
    covariance_type="full",
    n_init=10,
    random_state=42
)

labels = gmm.fit_predict(X_scaled)
probabilities = gmm.predict_proba(X_scaled)

print("GMM trained successfully.")
print("Converged:", gmm.converged_)
print("Number of EM iterations:", gmm.n_iter_)
print("Silhouette Score:", round(silhouette_score(X_scaled, labels), 4))
print("Adjusted Rand Index:", round(adjusted_rand_score(y, labels), 4))
print("Average Log-Likelihood:", round(gmm.score(X_scaled), 4))
```

```python

# Train the Gaussian Mixture Model first
gmm = GaussianMixture(
    n_components=3,
    covariance_type="full",
    n_init=10,
    random_state=42
)

labels = gmm.fit_predict(X_scaled)
probabilities = gmm.predict_proba(X_scaled)

# Display the learned component means and weights
component_summary = pd.DataFrame(
    gmm.means_,
    columns=X.columns
)

component_summary.insert(
    0, "Component Weight", gmm.weights_
)

print("Learned component parameters:")
display(component_summary.round(3))

print("Membership probabilities for the first 10 samples:")
display(pd.DataFrame(
    probabilities[:10],
    columns=[f"Component {i + 1}" for i in range(3)]
).round(4))
```

```python

# Select the number of Gaussian components using AIC and BIC

component_counts = range(1, 8)
selection_results = []

for k in component_counts:
    model = GaussianMixture(
        n_components=k,
        covariance_type="full",
        n_init=5,
        random_state=42
    )

    model.fit(X_scaled)

    selection_results.append({
        "Components": k,
        "AIC": model.aic(X_scaled),
        "BIC": model.bic(X_scaled)
    })

selection_df = pd.DataFrame(selection_results)

display(selection_df.round(2))

# Plot AIC and BIC
plt.figure(figsize=(8, 5))

plt.plot(
    selection_df["Components"],
    selection_df["AIC"],
    marker="o",
    label="AIC"
)

plt.plot(
    selection_df["Components"],
    selection_df["BIC"],
    marker="s",
    label="BIC"
)

plt.xlabel("Number of Gaussian Components")
plt.ylabel("Information Criterion")
plt.title("GMM Model Selection Using AIC and BIC")
plt.xticks(list(component_counts))
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

print(
    "Components with lowest AIC:",
    int(selection_df.loc[selection_df["AIC"].idxmin(), "Components"])
)

print(
    "Components with lowest BIC:",
    int(selection_df.loc[selection_df["BIC"].idxmin(), "Components"])
)
```
